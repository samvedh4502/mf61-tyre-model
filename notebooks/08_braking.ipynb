```python
import importlib
import numpy as np
from tir_reader import load_tir
import tyre_model as tm
importlib.reload(tm)

tyre = load_tir("data/sample.tir")


def best_braking_slip(tyre, Fz, alpha=0.0):
    kappa = np.linspace(-0.5, 0, 5001)
    Fx, Fy = tm.combined_force(tyre, kappa, alpha, Fz)
    i = np.argmin(Fx)
    return kappa[i], Fx[i], Fy[i]


for Fz in [2000, 4000, 6000, 8000]:
    k_best, Fx_best, _ = best_braking_slip(tyre, Fz)
    Fx_locked, _ = tm.combined_force(tyre, -1.0, 0.0, Fz)
    print(f"Fz {Fz}: best slip {k_best:.3f}, Fx {Fx_best:.0f} N, "
          f"mu {-Fx_best / Fz:.3f}, locked wheel keeps {Fx_locked / Fx_best:.0%}")
```

```python
for a_deg in [0, 2, 4, 8]:
    k_best, Fx_best, Fy_best = best_braking_slip(tyre, 4000, np.radians(a_deg))
    _, Fy_free = tm.combined_force(tyre, 0.0, np.radians(a_deg), 4000)
    print(f"alpha {a_deg} deg: best slip {k_best:.3f}, Fx {Fx_best:.0f} N, "
          f"Fy {Fy_best:.0f} N (free rolling {Fy_free:.0f} N)")
```
